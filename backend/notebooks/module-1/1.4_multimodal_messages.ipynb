## Text input

```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model

model = init_chat_model("gemini-2.5-flash-lite", model_provider="google_genai")

# from langchain_google_genai import ChatGoogleGenerativeAI

# model = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")


agent = create_agent(
    model=model,
    system_prompt="你是一名科幻作家，应用户要求打造一座都城",
)
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content=[
    {"type": "text", "text": "月球的首都在哪里"}
])

response = agent.invoke(
    {"messages": [question]}
)

print(response['messages'][-1].content)
```

## 图像输入

#### 多模态，就是大模型不仅可以识别文字，也可以去识别图像或者是声音，把图像，声音转换成信息的描述。


<img src="./resources/moon.png" width="400" style="display:block; margin-left:0;">

#### 例如，我们现在要让大模型帮我描述一张这张图片。

```python
from ipywidgets import FileUpload
from IPython.display import display

uploader = FileUpload(accept='.png', multiple=False)
display(uploader)
```

```python
print(uploader.value)
```

```python
import base64

# Get the first (and only) uploaded file dict
uploaded_file = uploader.value[0]

# This is a memoryview
content_mv = uploaded_file["content"]

# Convert memoryview -> bytes
img_bytes = bytes(content_mv)  # or content_mv.tobytes()

# Now base64 encode
img_b64 = base64.b64encode(img_bytes).decode("utf-8")
```

```python
multimodal_question = HumanMessage(content=[
    {"type": "text", "text": "请描述这个都城"},
    {"type": "image", "base64": img_b64, "mime_type": "image/png"}
])

response = agent.invoke(
    {"messages": [multimodal_question]}
)

print(response['messages'][-1].content)
```

## 音频输入

```python
import sounddevice as sd
from scipy.io.wavfile import write
import base64
import io
import time
from tqdm import tqdm

# Recording settings
duration = 10  # seconds
sample_rate = 44100

print("Recording...")
audio = sd.rec(int(duration * sample_rate), samplerate=sample_rate, channels=1)
# Progress bar for the duration
for _ in tqdm(range(duration * 10)):   # update 10× per second
    time.sleep(0.1)
sd.wait()
print("Done.")

# Write WAV to an in-memory buffer
buf = io.BytesIO()
write(buf, sample_rate, audio)
wav_bytes = buf.getvalue()

aud_b64 = base64.b64encode(wav_bytes).decode("utf-8")
```

```python
agent = create_agent(
    model=model,
)

multimodal_question = HumanMessage(content=[
    {"type": "text", "text": "告诉我这段录音的内容"},
    {"type": "audio", "base64": aud_b64, "mime_type": "audio/wav"}
])

response = agent.invoke(
    {"messages": [multimodal_question]}
)

print(response['messages'][-1].content)
```
