# 1. 变量与赋值（课程核心基础 · 15分钟）

所有AI调用都需要：
- 存储用户问题

**变量 = 程序里的“容器”**

# 2. Python四大基础数据类型（AI开发必备 · 10分钟）
## 为什么要学这一节？
大模型所有交互数据，**归根结底只有4种类型**：
1. 字符串：Prompt、回答、文本内容
2. 整数/小数：模型参数、费用、Token数量
3. 布尔值：判断开关、是否出错、是否流式输出

## 知识点讲解
- `str` 字符串：**所有文字内容**
- `int` 整数：**整数数字**
- `float` 浮点数：**小数、精度参数**
- `bool` 布尔值：`True` / `False`

```python
# 字符串 String：所有Prompt和回答都是字符串
prompt = "请解释什么是大模型"

# 整数 Int：最大生成长度、轮次、ID
max_length = 2048

# 浮点数 Float：温度、随机度、学习率
temperature = 0.75

# 布尔 Bool：状态判断
enable_log = True

# type() 查看数据类型（调试必备）
print("prompt 类型:", type(prompt))
print("max_length 类型:", type(max_length))
print("temperature 类型:", type(temperature))
print("enable_log 类型:", type(enable_log))
```

# 3. 字符串操作（Prompt组装核心技能 · 10分钟）
## 为什么要学这一节？
**90%的AI开发工作都在操作字符串**
- 拼接系统提示词 + 用户问题
- 清洗AI返回的多余换行、空格
- 截取部分回答、过滤无效字符

## 常用方法
- 字符串拼接 `+`
- 去除首尾空格 `.strip()`
- 替换字符 `.replace()`

```python
# 1. 标准大模型 Prompt 拼接方式
system_prompt = "你是专业AI助手，回答简洁准确。"
user_input = "什么是 LangChain？"

# 组合成完整提问
full_prompt = system_prompt + user_input
print("完整Prompt内容：")
print(full_prompt)
print("-" * 50)

# 2. 清洗模型返回脏数据（真实场景高频）
ai_raw_answer = "   DeepSeek 是一款优秀的通用大语言模型。\n\n功能强大   "

# 去除首尾空格
clean_answer = ai_raw_answer.strip()
# 去除换行符
clean_answer = clean_answer.replace("\n", "")

print("清洗后的模型回答：")
print(clean_answer)
```

# 4. 列表 List（多轮对话存储结构 · 15分钟）
## 为什么要学这一节？
**所有大模型平台的多轮对话，底层全是列表！**
- 列表用来存放**多轮历史对话**
- 顺序固定、可追加、可遍历

## 知识点
- 列表 `[]` 有序、可重复
- `.append()` 向后添加一轮对话
- 循环遍历所有对话记录

```python
# 对话历史是空列表
chat_history = []

# 追加用户对话
chat_history.append({"role": "user", "content": "你好"})
# 追加模型回复
chat_history.append({"role": "assistant", "content": "你好，我是 DeepSeek 大模型"})
# 追加第二轮提问
chat_history.append({"role": "user", "content": "LangChain 有什么作用？"})

# 遍历输出完整对话
print("=== 完整多轮对话历史 ===")
for msg in chat_history:
    print(msg["role"], "：", msg["content"])
```

**大模型API交互，100%依赖字典**
- 模型名称、温度、最大长度、Prompt、流式开关
- 所有参数都是 `key: value` 结构

字典是AI开发**最重要、最高频**的数据结构

## 知识点
- 字典 `{}` 键值对存储

# 6. 函数 Function（代码封装、一键调用AI · 10分钟）
## 为什么要学这一节？
如果不写函数：
- 每次调用AI都要写十几行重复代码
- 无法复用、无法统一管理

**函数 = 把重复代码打包成一个工具**

## 知识点
- `def` 定义函数
- 参数输入、结果返回
- 一次定义、无限调用

```python
# 封装通用问答函数
def ai_chat(question):
    """
    简易AI对话函数
    :param question: 用户输入问题
    :return: 模型模拟回答
    """
    answer = f"【AI收到提问】{question} → 回答：问题已解析完成"
    return answer

# 多次调用、无需重复写代码
print(ai_chat("什么是大模型API调用？"))
print(ai_chat("Python为什么适合AI开发？"))
print(ai_chat("如何使用DeepSeek模型？"))
```

# 7. 文件读写（对话日志持久化 · 5分钟）
## 为什么要学这一节？
真实项目需要：
- 保存每日对话记录
- 保存Prompt模板
- 保存模型输出结果

```python
# 写入对话日志到本地文件
log_content = """
用户：如何学习AI开发
模型：从Python基础 + 大模型API调用开始
"""

# 写入文件
with open("chat_log.txt", "w", encoding="utf-8") as f:
    f.write(log_content)

# 读取文件内容
with open("chat_log.txt", "r", encoding="utf-8") as f:
    result = f.read()

print("历史对话日志：")
print(result)
```

# 8. 异常处理（项目稳定必备 · 5分钟）
## 为什么要学这一节？
调用大模型极易报错：
- 网络超时
- API Key 错误
- 参数为空
- JSON解析失败

`try-except` 保证**程序不崩溃**

```python
def safe_chat(question):
    try:
        # 合法性校验
        if not question:
            return "错误：提问不能为空！"
        
        # 模拟AI回答
        return f"AI回答：{question}"
    
    except Exception as e:
        return f"系统异常：{str(e)}"

# 正常调用
print(safe_chat("大模型是什么？"))
# 异常调用（空输入）
print(safe_chat(""))
```

# 9. JSON数据转换（API通信核心 · 5分钟）
## 为什么要学这一节？
- Python只能识别字典
- 网络传输**只能识别JSON字符串**
- 所有大模型API：字典 ↔ JSON互转

```python
import json

# 1. 字典转JSON（发送请求使用）
request_data = {
    "model": "deepseek-chat",
    "prompt": "你好",
    "temperature": 0.6
}

json_data = json.dumps(request_data, ensure_ascii=False, indent=2)
print("JSON请求体：")
print(json_data)

# 2. JSON转回字典（接收返回结果使用）
dict_data = json.loads(json_data)
print("\n解析后的模型名称：", dict_data["model"])
```

# 10. 终极实战：LangChain 调用 DeepSeek 大模型（20分钟）
## 本节价值
Use official `init_chat_model` to initialize and call DeepSeek chat model quickly.

## 前置准备
1. Apply for your valid DeepSeek API Key in advance
2. Install LangChain and DeepSeek related dependencies

```python

import os
from langchain.chat_models import init_chat_model

# Initialize DeepSeek model (official standard code)
model = init_chat_model(model="deepseek-chat", model_provider="deepseek")

# Send question and get response
user_prompt = "用通俗易懂的话解释什么是大模型？"
response = model.invoke(user_prompt)

# Print output
print("【DeepSeek Model Response】")
print(response.content)
```
